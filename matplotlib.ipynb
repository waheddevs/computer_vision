```python
import matplotlib.pyplot as plt
import numpy as np
```

Basic line plot

```python
x = np.array([1, 2, 3, 4])
y = np.array([25, 10, 5, 40])

plt.plot(x, y)
plt.title('Basic line plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.show()
```

Line styles and colors

```python
x = np.array([5, 12, 50, 30])
y = np.array([2, 80, 55, 40])

plt.plot(y, linestyle='dotted', color='orange', marker='o')
plt.title('Styled Line Example', color='orange')
plt.show()
```

Multiple lines in one graph with Numpy and Matplotlib

```python
from matplotlib.lines import lineStyles
x = np.array([5, 12, 50, 30])
y1 = np.array([28, 10, 75, 30])
y2 = np.array([8, 42, 5, 20])

plt.plot(x, y1, label='Line 1', color='red', marker='s')
plt.plot(x, y2, label='Line 2', color='blue',linestyle='--', marker='o')
plt.title('Multiple lines example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.show()
```

Barchart with Numpy and Matplot

```python
x = np.array(['Salary', 'Expence', 'Charity', 'Food'])
y = np.array([2, 80, 55, 40])

plt.bar(x, y, color='orange')
plt.title('Barchart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

Scatter Plot with Numpy and Matplot

```python
x = np.array([5, 12, 50, 30, 35])
y = np.array([28, 10, 75, 30, 45])

plt.scatter(x, y, color='purple', s=150)
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.xlabel('Y')
plt.show()
```

Histogram with Numpy and Matplot

```python
data = np.array([5, 12, 50, 30, 35, 28, 10, 75, 30, 45])
plt.hist(data, bins=5, color='skyblue', edgecolor='black')
plt.title('Histogram Example')
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.show()
```

Subplots(Multiple Graphs Together)

```python
x = np.array([1, 5, 8, 10])
y1 = np.array([12, 15, 20, 25])
x2 = np.array([50, 55, 70, 90])

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='blue')
plt.title('Plot 2')

plt.tight_layout()
plt.show()
```
