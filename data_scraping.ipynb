```python
import requests
import time
from pathlib import Path
from io import BytesIO
from PIL import Image

folder = Path("/content/plov_commons")
folder.mkdir(parents=True, exist_ok=True)

session = requests.Session()
session.headers["User-Agent"] = "PlovDatasetStudy/1.0 (educational image collection)"

response = session.get(
    "https://commons.wikimedia.org/w/api.php",
    params={
        "action": "query",
        "format": "json",
        "generator": "search",
        "gsrsearch": "plov",
        "gsrnamespace": 6,
        "gsrlimit": 40,
        "prop": "imageinfo",
        "iiprop": "url|mime",
        "iiurlwidth": 960,
    },
    timeout=30,
)
response.raise_for_status()
data = response.json()

if "error" in data:
    raise RuntimeError(data["error"])

pages = data.get("query", {}).get("pages", {})
count = 0

for page in pages.values():
    info = page.get("imageinfo", [{}])[0]
    if info.get("mime") not in {"image/jpeg", "image/png", "image/webp"}:
        continue

    url = info.get("thumburl") or info.get("url")
    if not url:
        continue

    try:
        result = session.get(url, timeout=30)
        result.raise_for_status()

        with Image.open(BytesIO(result.content)) as img:
            img = img.convert("RGB")
            if min(img.size) < 200:
                continue

            img.save(folder / f"plov_{page['pageid']}.jpg")

        count += 1
        print(f"{count}. {page['title']}")

    except (requests.RequestException, OSError) as error:
        print(f"Yuklanmadi: {page['title']} — {error}")

    time.sleep(1)
    if count >= 80:
        break

print(f"\nSaqlandi: {count} ta rasm")
print(f"Papka: {folder}")
```

```python
import requests
import time
from pathlib import Path
from io import BytesIO
from PIL import Image

folder = Path("/content/kebab_commons")
folder.mkdir(parents=True, exist_ok=True)

session = requests.Session()
session.headers["User-Agent"] = "KebabDatasetStudy/1.0 (educational image collection)"

response = session.get(
    "https://commons.wikimedia.org/w/api.php",
    params={
        "action": "query",
        "format": "json",
        "generator": "search",
        "gsrsearch": "kebab",
        "gsrnamespace": 6,
        "gsrlimit": 150,
        "prop": "imageinfo",
        "iiprop": "url|mime",
        "iiurlwidth": 960,
    },
    timeout=30,
)
response.raise_for_status()
data = response.json()

if "error" in data:
    raise RuntimeError(data["error"])

pages = data.get("query", {}).get("pages", {})
count = 0
limit = 100

for page in pages.values():
    file_path = folder / f"kebab_{page['pageid']}.jpg"

    if file_path.exists():
        continue

    info = page.get("imageinfo", [{}])[0]
    if info.get("mime") not in {"image/jpeg", "image/png", "image/webp"}:
        continue

    url = info.get("thumburl") or info.get("url")
    if not url:
        continue

    try:
        result = session.get(url, timeout=30)
        result.raise_for_status()

        with Image.open(BytesIO(result.content)) as img:
            img = img.convert("RGB")
            if min(img.size) < 200:
                continue

            img.save(file_path)

        count += 1
        print(f"{count}. {page['title']}")

    except (requests.RequestException, OSError) as error:
        print(f"Yuklanmadi: {page['title']} — {error}")

    time.sleep(1)
    if count >= limit:
        break

print(f"\nYangi saqlandi: {count} ta rasm")
print(f"Papka: {folder}")
```
